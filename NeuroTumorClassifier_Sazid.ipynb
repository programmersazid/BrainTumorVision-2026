```python
# Dataset Training Testing file path set
Train = r"C:\Users\kioskUser0\Documents\ModabbirHasanSazid\CranialMRI_ScanArchive\Training"
Test = r"C:\Users\kioskUser0\Documents\ModabbirHasanSazid\CranialMRI_ScanArchive\Testing"

# Training Data Preparation

train_data_paths = []
train_data_labels = []

for label in os.listdir(Train):
    for image in os.listdir(os.path.join(Train, label)):
         train_data_paths.append(os.path.join(Train, label, image))
         train_data_labels.append(label)

train_data_paths , train_data_labels = shuffle(train_data_paths, train_data_labels)

# Test Data Preparation 
test_data_paths = []
test_data_labels = []

for label in os.listdir(Test):
    for image in os.listdir(os.path.join(Test, label)):
         test_data_paths.append(os.path.join(Test, label, image))
         test_data_labels.append(label)

test_data_paths , test_data_labels = shuffle(test_data_paths, test_data_labels)





```
